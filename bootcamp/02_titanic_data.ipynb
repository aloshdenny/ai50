```python
#@title ▶ run me first (loads the titanic data)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="white")
titanic = sns.load_dataset("titanic")
print(f"loaded: {len(titanic)} passengers, {titanic.shape[1]} columns ✅")
```

## 1 · look at it

```python
#@title show the first few rows
rows = 6 #@param {type:"slider", min:1, max:30, step:1}
titanic[["survived", "pclass", "sex", "age", "fare", "embark_town"]].head(rows)
```

each **row** is a passenger. each **column** is one detail. `survived` is the **label**: 1 = survived, 0 = didn't.

the same thing in code is one line. press ▶:

```python
titanic.head()
```

## 2 · how messy is it?

```python
#@title count the missing values
missing = titanic.isna().sum()
missing = missing[missing > 0].sort_values()
missing.plot.barh(color="black", figsize=(8, 3), title="missing values per column")
plt.show()
print(missing)
```

177 passengers have no age. 688 have no deck. that's real data: **messy**.

## 3 · who survived?

```python
#@title survival rate, split by...
split_by = "sex" #@param ["sex", "class", "who", "alone", "embark_town", "deck"]
rates = titanic.groupby(split_by, observed=True)["survived"].mean().mul(100).round(1)
ax = rates.plot.bar(color="black", figsize=(7, 4), rot=0, title=f"% who survived, by {split_by}")
ax.bar_label(ax.containers[0], fmt="%.0f%%")
plt.show()
```

try `sex` first (74% vs 19%, same as yesterday), then `class`, then `who`. which split tells the biggest story?

## 4 · ages on board

```python
#@title age histogram
bins = 20 #@param {type:"slider", min:5, max:60, step:5}
plt.figure(figsize=(8, 4))
sns.histplot(data=titanic, x="age", hue="alive", bins=bins, multiple="stack", palette=["#c7c7c7", "black"])
plt.title("ages of passengers (black = survived)")
plt.show()
```

notice the little black bump at the far left: **children**. 'women and children first' is right there in the data.

## 5 · the £512 ticket

```python
#@title fares: spot the outlier
plt.figure(figsize=(8, 2.5))
sns.boxplot(x=titanic["fare"], color="#f2f2f2")
plt.title("ticket prices (£)")
plt.show()
print(f"average fare: £{titanic.fare.mean():.2f}")
print(f"highest fare: £{titanic.fare.max():.2f}")
titanic[titanic.fare == titanic.fare.max()][["sex", "age", "pclass", "fare", "survived"]]
```

the dots far to the right are **outliers**. real people, very unusual tickets.

## 6 · find people like you

```python
#@title filter the passengers
sex = "female" #@param ["female", "male"]
passenger_class = 3 #@param [1, 2, 3]
min_age = 18 #@param {type:"slider", min:0, max:80, step:1}
max_age = 30 #@param {type:"slider", min:0, max:80, step:1}

people = titanic[(titanic.sex == sex) & (titanic.pclass == passenger_class) & titanic.age.between(min_age, max_age)]
print(f"{len(people)} passengers like that.")
if len(people):
    print(f"{people.survived.mean() * 100:.0f}% of them survived.")
```

## 7 · make your own chart

```python
#@title pick two columns
x = "pclass" #@param ["pclass", "sex", "age", "fare", "embark_town", "who", "deck"]
y = "fare" #@param ["fare", "age", "survived"]
chart = "bar" #@param ["bar", "box", "scatter"]
plt.figure(figsize=(8, 4))
if chart == "bar":
    sns.barplot(data=titanic, x=x, y=y, color="black", errorbar=None)
elif chart == "box":
    sns.boxplot(data=titanic, x=x, y=y, color="#f2f2f2")
else:
    sns.scatterplot(data=titanic, x=x, y=y, hue="alive", palette=["#c7c7c7", "black"])
plt.title(f"{y} by {x}")
plt.show()
```
