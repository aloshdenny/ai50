## 2 · lists: lots of things in one box

```python
friends = ["anu", "ravi", "meera", "arjun"]   # add a name, keep the quotes and commas

print("how many friends:", len(friends))
print("first friend:", friends[0])     # computers count from 0. yes, really.
print("last friend:", friends[-1])
```

## 3 · loops: do it again, for each thing

```python
for friend in friends:
    print("good morning", friend, "☀️")
```

the indented line runs **once per friend**. add a fifth friend to the list above, run both cells again.

```python
number = 7    # change me

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

a times table, without writing it out ten times. that's the whole point of loops.

```python
rows = 5        # change me
food = "🍕"     # try "🍵" or "🥟"

for i in range(1, rows + 1):
    print(food * i)
```

## 4 · if: make a decision

```python
marks = 72    # change me

if marks >= 40:
    print("pass 🎉")
else:
    print("try again next time")
```

remember the umbrella tree from day 3? here it is in python:

```python
cloudy = True               # True or False
forecast_says_rain = False  # True or False

if cloudy:
    if forecast_says_rain:
        print("take the umbrella ☂️")
    else:
        print("risk it")
else:
    print("no umbrella 😎")
```

a decision tree is just `if`s inside `if`s. the clever part of machine learning is that the **computer writes these ifs itself**, from data.

## 5 · a loop with an if inside

```python
for number in range(1, 16):
    if number % 3 == 0:          # % means "remainder". 9 % 3 is 0, so 9 divides by 3
        print(number, "🍵 chai break")
    else:
        print(number)
```

## 6 · dictionaries: a list with labels

```python
menu = {"chai": 15, "samosa": 20, "vada pav": 25, "filter coffee": 30}   # add your own item

for item, price in menu.items():
    print(f"{item:15} ₹{price}")
```

## 7 · functions: give a recipe a name

```python
def bill(cups, price=15):
    return cups * price

print(bill(2))
print(bill(10))
print(bill(3, price=40))    # fancy café
```

## 8 · numpy: maths on a whole list at once

```python
import numpy as np

marks = np.array([72, 45, 88, 91, 38, 66])   # change these

print("average:", marks.mean())
print("highest:", marks.max())
print("everyone +5 grace marks:", marks + 5)
```

no loop needed. numpy does the whole list in one go. this is what machine learning libraries are built on.

## 9 · your first chart

```python
import matplotlib.pyplot as plt

days = ["mon", "tue", "wed", "thu", "fri"]
cups = [2, 3, 5, 4, 1]        # how much chai did you drink?

plt.bar(days, cups, color="black")
plt.title("cups of chai this week")
plt.show()
```
